# Calcul des moyennes des UE

```python
somme=0
coeff=0
for i in range (101,116):
    n=float(input("indiquez votre note dans la ressource R{}".format(i)))
    c=int(input("indiquez votre coeff dans la ressource R{}".format(i)))
for i in range (1,5):
    n=float(input("indiquez votre note dans la ressource SAE{}".format(i)))
    c=int(input("indiquez votre coeff dans la ressource SAE{}".format(i)))
    somme+=n*c
    coeff+=c
moy=somme/coeff
print("la moyenne de votre UE est donc:",moy)
```
