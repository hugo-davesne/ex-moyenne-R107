```python
from tkinter import *
from tkinter import ttk

#création d'une fenetre
fenetre=Tk()

#creation d'un cadre dans la fenetre et d'une grille
frm=ttk.Frame(fenetre)
frm.grid()

#creation d'un texte en vue d'un placement d'un cadre
objet=ttk.Label(frm,text="première valeur")

#affichage du texte dans le cadre, selon les coordonnées précisées dans la grille
objet.grid(column=0,row=0, padx=20,pady=20)

fenetre.mainloop()
```

```python
from tkinter import *
from tkinter import ttk

fenetre=Tk()

frm=ttk.Frame(fenetre)
frm.grid()

objet1=ttk.Label(frm,text="première valeur")
objet1.grid(column=0,row=0, padx=20,pady=20)

objet2=ttk.Label(frm,text="deuxième valeur")
objet2.grid(column=0,row=1, padx=20,pady=20)

valeur1 = ttk.Entry(frm)
valeur1.grid(column = 1, row = 0, padx = 20)

valeur2 = ttk.Entry(frm)
valeur2.grid(column = 1, row = 1, padx = 20)

def calcul() :
    somme = int(valeur1.get()) + int(valeur2.get())
    resultat = ttk.Label(frm, text = f"la somme des deux valeurs est {somme}")
    resultat.grid(column = 2, row=1,)

bouton = ttk.Button(frm, text = "valider les saisies", command = calcul)
bouton.grid (column = 2, row = 0, padx = 30)

fenetre.mainloop()
```
