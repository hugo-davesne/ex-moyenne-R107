# Calcul des moyennes des UE V2

```python
notes = (11,0,5,15,18,14,14,14,14,17,13,15,12,14,12,14,13,11,19,7,1)
coeff = (10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7)
somme_notes = 0
somme_coeff = 0

for i in range(len(notes)):
    somme_notes += notes[i] * coeff[i]
    somme_coeff += coeff[i]
moyenne = somme_notes / somme_coeff
print("Votre moyenne est donc de",moyenne,)
```
